```python
# Creating Series pd
Data = [ 2,4,6,8]
s = pd.Series(Data)
print(s)
```

```python
Data = [1,2,3,4]
s = pd.Series(Data)
print(s)
```

```python
Data = [1,2,3,4,5]
i=['a','b','c','d','e']
s = pd.Series(Data,index =i)
print(s)

#updating
s['a'] = 10
print(s)
s['b'] =50
print(s)

#inserting new data(directly,concat)
s["f"] = 80
print(s)

# remove (drop method)
s = s.drop("a")
print(s)
```

```python
# Creating a Series from a Dictionary
s2 =pd.Series({'x': 100, 'y': 200, 'z': 300})
print(s2)
```

```python
# Accessing elements
print(s2[0])
print(s2[0:2])
print(s2[1:])
print(s2[:2])
print(s2[2])

```
